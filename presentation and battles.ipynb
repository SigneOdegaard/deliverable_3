# Character class  
Each character has a name, health points (HP), maximum health points, and damage.

### Attributes
    name: The character's name.

    _hp: The character's current health points (private attribute).

    hp_max: The maximum health points.

    damage: The damage dealt when attacking.

### Methods 
    introduce(): Introduces the character.

    take_damage(amount): Reduces the character's HP.

    heal(amount): Restores the character's HP.

    attack(target): Attacks another character.

    __str__(): Returns a readable string representation of the character.

    __repr__(): Returns a representation of the character object.


The class has a hit point property (allows us to access the current health points.)
Also an is_alive property (checks whether the character is alive.)

# Creating a Character Object

We create an object called hero from the Character class. The character is given a name, maximum HP and damage.

```python
# Create object 
hero = Character("Aragorn", 100, 15)

# Display attributes of the object
print("Name:", hero.name)
print("Current HP:", hero.hp)
print("Maximum HP:", hero.hp_max)
print("Damage:", hero.damage)
print("Is alive:", hero.is_alive)

# Demonstrate introduce()
hero.introduce()
```

# Dice class
the Dice class represents a dice with a set number of sides that randomly rolls a number that decides the amount of damage dealt by the fighters

### Attributes
    sides (int): The number of sides on the dice. The default is 6, and the minimum allowed value is 2.

### Methods
    inii(sides=6): Initializes a Dice object with the specified number of sides.
    roll(): Rolls the dice and returns a random integer between 1 and the number of sides.
    repr(): Returns a string representation of the object, such as Dice(sides=6).
    call(): Allows the Dice object to be called like a function, producing a dice roll without explicitly calling roll().

```python
# demonstrate Dice class
d12 = Dice(12)

#attributes and repr()
print("number of sides:", d12.sides)

#Demonstrate multiple rolls
print("Five rolls with d12:")
for i in range(5):
    print(d12.roll())
```

# Weapon Class
The weapon class represents an item that the characters can equip. each weapon uses a dice object to determine damage dealt during attacks

### Attributes 
    name: the name of the weapon
    damage_dice: the dice used to determine the damage

### Methods 
    init(): initialize the wepon
    roll_damage(): rolls the dice to decide the damage dealt

```python
# Demonstrate weapon class
sword = Weapon("mjolnir", Dice(8))
print("Weapon:", sword.name)

print("five attacks from weapon:")
for i in range(5):
    print(sword.roll_damage())
    
```

# Warrior class

Warrior class is a subclass of Character. 
It inherits the attributes and methods from the Character class. It also has its own additional attribute, armor, and overrides the take_damage() method to reduce incoming damage based on its armor value.

```python
# Import Character and Warrior from the character.py file

from character import Character, Warrior
```

## Creating an warrior object

Create an object called warrior from the Warrior class. 
The warrior is given a 
--> name
--> maximum HP
--> damage 
--> armor value

--> name: The warrior's name.

--> hp: The warrior's current health points.

--> hp_max: The maximum health points.

--> damage: The damage dealt when attacking.


```python
# Display attributes: 

print("Name:", warrior.name)
print("Current HP:", warrior.hp)
print("Maximum HP:", warrior.hp_max)
print("Damage:", warrior.damage)
print("Armor:", warrior.armor)
print("Is alive:", warrior.is_alive)
```

--> introduce(): Introduces the warrior.

# Mage class

Subclass of the character class.
It inherits the attributes and methods from the Character class, but also has its own mana attributes.

```python
# Import and create a Mage object.

from character import Character, Mage

mage = Mage("Merlin", 100, 15)

# --> create a Mage named Merlin with 100 HP and 15 base damage. Mana automatically starts at 30 
```

## Attributes: 
--> mana_max: The maximum amount of mana the Mage can have.
--> name: Inherited from the Character class.
--> hp: Inherited property showing the Mage's current health points.
--> hp_max: Inherited from the Character class.
--> damage: Inherited from the Character class.
--> is_alive: Inherited property that indicates whether the Mage is alive.

```python
# Display attributes: 

print("Name:", mage.name)
print("Current HP:", mage.hp)
print("Maximum HP:", mage.hp_max)
print("Damage:", mage.damage)
print("Current mana:", mage.mana)
print("Maximum mana:", mage.mana_max)
print("Is alive:", mage.is_alive)
```

## Metoder used in Mage class

--> attack(target): Overrides the attack() method from Character. Uses a magical attack that deals double damage when the Mage has at least 10 mana. Otherwise, it performs a standard attack and restores 5 mana.

--> take_damage(amaount): Inherited from Character. Reduces the Mage's HP when taking damage.
--> heal(amount): Inherited from Character. Restores the Mage's HP.
--> introduce: Inherited from Character. Prints an introduction.
--> __str()__: Inherited from Character. Returns a readable string representation of the Mage.
--> __repr()__: Inherited from Character. Returns a representation of the Mage object

# Battle without weapons – Warrior vs. Mage

The Warrior uses armor to reduce incoming damage, while the Mage uses mana to perform magical attacks.

```python
from character import Warrior, Mage
from battle import Battle

warrior = Warrior("Thor", hp_max=120, damage=15, armor=5)
mage = Mage("Merlin", hp_max=100, damage=12)

battle = Battle(warrior, mage)
battle.simulate_battle()
```

# Battle with weapons – Warrior vs. Mage

The weapons use dice to determine the damage dealt during attacks.

```python
from items import Weapon
from dice import Dice

warrior_weapon = Weapon("Sword", Dice(12))
mage_weapon = Weapon("Magic Staff", Dice(10))

warrior = Warrior(
    "Thor",
    hp_max=120,
    damage=15,
    armor=5,
    weapon=warrior_weapon
)

mage = Mage(
    "Merlin",
    hp_max=100,
    damage=12,
    weapon=mage_weapon
)

battle = Battle(warrior, mage)
battle.simulate_battle()
```
